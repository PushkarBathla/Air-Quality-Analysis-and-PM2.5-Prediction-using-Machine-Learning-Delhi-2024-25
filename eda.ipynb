# Dataset contains air quality parameters with timestamp, location, and pollutant values. Initial inspection shows presence of missing values and mixed data types.

# 

```python
cols_to_drop = [
    'Xylene (µg/m³)', 'O Xylene (µg/m³)',
    'Eth-Benzene (µg/m³)', 'MP-Xylene (µg/m³)',
    'RF (mm)', 'VWS (m/s)'
]

df = df.drop(columns=cols_to_drop)
```

```python
df = df.dropna(subset=['Timestamp', 'City', 'Station Name'])
```

```python
df = df.sort_values(by='Timestamp')
df
```

# Dropped all Empty Unneccessary Columns that were not required and made data sorted according to the Timestamp

#

```python
# Clean column names
df.columns = df.columns.str.strip().str.lower()

# Convert timestamp
df['timestamp'] = pd.to_datetime(df['timestamp'], errors='coerce')

# Define categorical columns
categorical_cols = ['station id', 'state', 'city', 'station name']

# Clean ONLY numeric columns
for col in df.columns:
    if col not in categorical_cols + ['timestamp']:
        df[col] = df[col].replace(r'[^\d.-]', '', regex=True)

# Convert numeric columns
numeric_cols = [col for col in df.columns 
                if col not in categorical_cols + ['timestamp']]

df[numeric_cols] = df[numeric_cols].apply(pd.to_numeric, errors='coerce')
df
```

# Special characters and units were removed from numeric columns to ensure proper numerical conversion.

```python
df = df.set_index('timestamp')
df = df[~df.index.isna()]
df = df.sort_index()
df
```

```python
df[numeric_cols] = df[numeric_cols].interpolate(method='time')
df[numeric_cols] = df[numeric_cols].bfill().ffill()
df
```

# Missing values were handled using time-based interpolation, followed by forward and backward filling to ensure no gaps remain.

# 

```python
for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    df[col] = df[col].clip(lower, upper)
```

# Outliers were detected and handled using the IQR (Interquartile Range) method to reduce the effect of extreme values.

```python
df.to_csv("cleaned_air_quality.csv", index=True)
```

```python
import seaborn as sns

df = pd.read_csv("cleaned_air_quality.csv")
df['timestamp'] = pd.to_datetime(df['timestamp'])
df = df.set_index('timestamp')
```

```python
numeric_cols = df.select_dtypes(include=['number']).columns
print(numeric_cols)
print(df.columns)
```

```python
stations = df['station name'].unique()

for station in stations:
    station_df = df[df['station name'] == station].sample(5000)
    
    station_df[numeric_cols].hist(figsize=(12,10), bins=30)
    
    plt.suptitle(f"Distribution of Air Pollutants - {station}")
    plt.tight_layout()
    plt.show()
```
